    Activation Function g(x) = {1 if x >= 0,

```python
def activationFunction(x):
    return 1 if x >= 0 else 0
```

```python
def netInputFunction(X, W):
    return np.dot(X, W)
    
```

```python
def perceptronUpdateRule(X, W, Y):
    for i in range(len(X)):
        y = activationFunction(netInputFunction(X[i], W))
        W = W + (Y[i] - y) * X[i]
    return W
```

```python
def approximation(X, W):
    return [activationFunction(netInputFunction(X[i], W)) for i in range(len(X))]
```

Training

```python
def train(X, Y, W):
    for i in range(len(Y)):
        W[i] = perceptronUpdateRule(X, W[i], Y[i])
    return W
```

```python
waights = np.array([np.random.rand(X.shape[1]) for i in range(Y.size)])
waights
```

```python
updatedWaights = train(X, Y, waights)
```
